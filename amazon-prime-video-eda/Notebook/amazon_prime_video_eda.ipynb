## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
#change column name in title
titles.rename(columns={'type':'show_type'}, inplace=True)
````

````python
#joining dataset of titles and credits keeping all titles and using left join
df = pd.merge(titles, credits, on='id', how='left')
df.shape
df.head()
````

````python
#removing duplicates
df.drop_duplicates(inplace=True)
````

````python
df.duplicated().sum()
````

````python
#handling null values
#age_certifications have so much null values
df['age_certification'] = df['age_certification'].fillna("Not Rated")
````

````python
#seasons - who have the zero season fill it with a zero
df['seasons'] = df['seasons'].fillna(0)
````

````python
#numeric columns
df['imdb_score'] = df['imdb_score'].fillna(df['imdb_score'].median())
df['tmdb_score'] = df['tmdb_score'].fillna(df['tmdb_score'].median())
df['imdb_votes'] = df['imdb_votes'].fillna(0)
df['tmdb_popularity'] = df['tmdb_popularity'].fillna(0)
````

````python
#Fix Data Types
df['release_year'] = df['release_year'].astype(int)
df['runtime'] = df['runtime'].fillna(0).astype(int)
df['seasons'] = df['seasons'].astype(int)
````

````python
#Clean TEXT Columns (Genres & Countries)
#for genres
df['genres'] = df['genres'].str.replace("[","")
df['genres'] = df['genres'].str.replace("]","")
df['genres'] = df['genres'].str.replace("'","")
````

````python
#same for production countries
df['production_countries'] = df['production_countries'].str.replace("[","")
df['production_countries'] = df['production_countries'].str.replace("]","")
df['production_countries'] = df['production_countries'].str.replace("'","")
````

````python
#handling missing names
#sometimes actors and directors names are missing
df['name'] = df['name'].fillna("Unknown")
df['character'] = df['character'].fillna("Unknown")
df['role'] = df['role'].fillna("Unknown")
````

````python
#Cleaning and removing unrealsic values
#outliers
df['runtime'].describe()
df = df[df['runtime'] > 0]

````

````python
#Create Decade Column
df['decade'] = (df['release_year']//10)*10

````

````python
#create runtime category
df['runtime_category'] = np.where(
    df['runtime'] < 60, "Short",
    np.where(df['runtime'] < 120, "Medium", "Long")
)
````

````python
#Seperate actors and directors
actors_df = df[df['role']=="ACTOR"]
directors_df = df[df['role']=="DIRECTOR"]
````

````python
#Final Cleaning
df.reset_index(drop=True, inplace=True)
````

### What all manipulations have you done and insights you found?

Data Wrangling we did so far
During the data wrangling process, several cleaning and preparation steps were performed to make the dataset suitable for analysis and visualization.

First, both datasets — titles and credits — were merged using the common id column so that information about shows, movies, actors, and directors could be analyzed together in a single dataframe.

After merging, duplicate rows were removed to avoid repeated information that could affect analysis results. Missing values were handled carefully. For example, missing age certifications were replaced with “Not Rated,” missing seasons were filled with 0 because movies do not have seasons, and missing rating values were replaced using median values to maintain realistic data distribution.

The text columns such as genres and production countries were cleaned by removing brackets and special characters so they could be used easily during analysis. Data types were also corrected by converting columns like release year, runtime, and seasons into proper numeric formats.

Some new features were created to make analysis easier. A decade column was added from release year to study trends over time, and a runtime category column (Short, Medium, Long) was created to understand content length patterns. Unrealistic values such as zero or negative runtime were removed to improve data quality.
Insights Found After Wrangling

After cleaning and organizing the data, several initial observations became clear. The dataset contains a mix of movies and TV shows, and many titles belong to multiple genres, showing strong content diversity on the platform. Filling missing seasons with zero helped clearly separate movies from shows.

The merged dataset also revealed that one title can have multiple actors and directors, which increased the number of rows after merging. Creating the decade column showed that content spans across many years, which will help analyze growth trends later.

Handling missing ratings ensured that no major gaps remain in numerical analysis, making visualizations more reliable. Overall, the wrangling process made the dataset structured, consistent, and ready for deeper exploratory data analysis and visualization.
